### Outliers

```python
results = []

cols = reading.select_dtypes(include=['float64', 'int64'])

for col in cols:
    q1 = reading[col].quantile(0.25)
    q3 = reading[col].quantile(0.75)
    iqr = q3 - q1
    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = reading[(reading[col] < lower_bound) | (reading[col] > upper_bound)]
    percent_outliers = (len(outliers) / len(reading)) * 100

    results.append({
        'Kolom': col,
        'Persentase Outliers': percent_outliers
    })

results_reading = pd.DataFrame(results)
results_reading.set_index('Kolom', inplace=True)

display(results_reading)
```

### Outliers

```python
results = []

cols = reading.select_dtypes(include=['float64', 'int64'])

for col in cols:
    q1 = reading[col].quantile(0.25)
    q3 = reading[col].quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outliers = reading[(reading[col] < lower_bound) | (reading[col] > upper_bound)]

    results.append({
        'Column': col,
        'Jumlah Outliers': outliers.shape[0]
    })

outliers_df = pd.DataFrame(results)
outliers_df
```

Hasil pengecekan outliers menggunakan metode Interquartile Range (IQR) menunjukkan bahwa variabel Age tidak memiliki outliers. Namun, pada variabel How many books did you read during last 12 months? ditemukan sejumlah 317 outliers. Hal ini mengindikasikan adanya perbedaan yang cukup ekstrem dalam jumlah buku yang dibaca responden, yang merupakan karakteristik umum pada data berbasis survei. Oleh karena itu, outliers tidak dihapus dan tetap dipertahankan untuk analisis selanjutnya.

### Data Construction

```python
def reading_level(x):
    if x == 0:
        return 'Tidak Membaca'
    elif x <= 5:
        return 'Rendah'
    elif x <= 15:
        return 'Sedang'
    else:
        return 'Tinggi'

reading['Reading_Level'] = reading['How many books did you read during last 12months?'].apply(reading_level)
```

```python
reading[['How many books did you read during last 12months?', 'Reading_Level']].head()
```

Data construction dilakukan dengan menambahkan variabel Reading_Level yang dibentuk dari variabel How many books did you read during last 12 months?. Penambahan variabel ini bertujuan untuk menyederhanakan data numerik yang memiliki rentang nilai cukup luas menjadi beberapa kategori yang lebih mudah dipahami, yaitu tidak membaca, tingkat rendah, sedang, dan tinggi.

Selain itu, pengelompokan ini membantu mengurangi pengaruh nilai ekstrem (outliers) tanpa menghilangkan data asli, sehingga pola kebiasaan membaca responden dapat dianalisis dengan lebih jelas. Variabel Reading_Level juga mempermudah proses analisis dan interpretasi hasil, khususnya dalam membandingkan kebiasaan membaca antar kelompok responden berdasarkan karakteristik tertentu.
